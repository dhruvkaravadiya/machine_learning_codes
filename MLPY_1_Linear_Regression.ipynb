## Theory

### 1. What is Linear Regression?
Linear Regression is a foundational supervised machine learning algorithm used to model the linear relationship between a continuous dependent target variable ($y$) and one or more independent predictor features ($X$). It assumes that the target variable can be modeled as a linear combination of the input variables plus a random error disturbance term ($\epsilon$).

### 2. Simple vs. Multiple Linear Regression
* **Simple Linear Regression:** Models the linear relationship between a single independent variable ($x$) and a single continuous target ($y$). In this laboratory experiment, years of professional experience ($x$) is used to predict salary ($y$).
* **Multiple Linear Regression:** Models the relationship between two or more independent variables ($x_1, x_2, \dots, x_p$) and a single continuous target ($y$).

### 3. Linear Regression Equation
* **Simple Linear Regression Equation:**
  $$y = eta_0 + eta_1 x + \epsilon \quad 	ext{or} \quad y = mx + c$$
  where $y$ is the target (Salary), $x$ is the feature (YearsExperience), $eta_0$ (or $c$) is the y-intercept, $eta_1$ (or $m$) is the slope/coefficient, and $\epsilon$ is the random error.
* **Multiple Linear Regression Equation:**
  $$y = eta_0 + eta_1 x_1 + eta_2 x_2 + \dots + eta_p x_p + \epsilon$$

### 4. Slope and Intercept
* **Intercept ($eta_0$ / $c$):** Represents the expected baseline value of $y$ when all predictor variables equal zero ($x = 0$). In this experiment, it represents the base salary of an individual with zero years of experience.
* **Slope ($eta_1$ / $m$):** Represents the rate of change in $y$ for every 1-unit increase in $x$. Here, it indicates the average annual salary increment per additional year of experience.

### 5. How the Model Learns the Best-Fit Line (Ordinary Least Squares - OLS)
The objective of linear regression is to find parameter estimates ($eta_0, eta_1$) that minimize the total prediction discrepancy. Ordinary Least Squares (OLS) minimizes the **Sum of Squared Residuals (SSR)**:
$$\min_{eta_0, eta_1} \sum_{i=1}^n (y_i - \hat{y}_i)^2 = \min_{eta_0, eta_1} \sum_{i=1}^n \left(y_i - (eta_0 + eta_1 x_i)
ight)^2$$
Setting partial derivatives with respect to $eta_0$ and $eta_1$ to zero yields the closed-form analytical solutions:
$$eta_1 = rac{\sum_{i=1}^n (x_i - ar{x})(y_i - ar{y})}{\sum_{i=1}^n (x_i - ar{x})^2}, \qquad eta_0 = ar{y} - eta_1 ar{x}$$
In vectorized matrix notation for multiple predictors:
$$oldsymbol{eta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}$$

### 6. MSE / Cost Function
The cost function measures average squared deviation across all $n$ observations:
$$J(eta_0, eta_1) = 	ext{MSE} = rac{1}{n} \sum_{i=1}^n (y_i - \hat{y}_i)^2 = rac{1}{n} \sum_{i=1}^n \left(y_i - (eta_0 + eta_1 x_i)
ight)^2$$

### 7. Residuals / Errors
For each observation $i$, the residual $e_i$ is defined as the deviation between actual and predicted target values:
$$e_i = y_i - \hat{y}_i$$
Residuals represent unexplained variation and form the basis for diagnostic tests and evaluation.

### 8. Train-Test Split
To evaluate generalization ability and prevent overfitting, the dataset is divided into:
* **Training Set (typically 80%):** Used exclusively by the algorithm to estimate model parameters ($eta_0, eta_1$).
* **Testing Set (typically 20%):** Held out during training and used strictly to evaluate model accuracy on unseen observations, preventing data leakage.

### 9. Basic Regression Evaluation Metrics
* **Mean Squared Error (MSE):** $	ext{MSE} = rac{1}{n}\sum_{i=1}^n (y_i - \hat{y}_i)^2$ (penalizes large errors heavily).
* **Root Mean Squared Error (RMSE):** $	ext{RMSE} = \sqrt{	ext{MSE}}$ (expressed in original currency units, e.g., ₹).
* **Mean Absolute Error (MAE):** $	ext{MAE} = rac{1}{n}\sum_{i=1}^n |y_i - \hat{y}_i|$ (linear average error magnitude, robust to outliers).
* **Coefficient of Determination ($R^2$):** $R^2 = 1 - rac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - ar{y})^2} = 1 - rac{	ext{SS}_{	ext{res}}}{	ext{SS}_{	ext{tot}}}$ (proportion of target variance explained by the model).

### 10. Important Assumptions of Linear Regression
1. **Linearity:** The relationship between predictor $X$ and target $y$ is strictly linear.
2. **Independence of Errors:** Residuals are independent across observations ($Cov(e_i, e_j) = 0$ for $i
eq j$), with no serial autocorrelation.
3. **Homoscedasticity:** Residual variance remains constant across all levels of the predictor ($Var(e_i) = \sigma^2$).
4. **Normality of Residuals:** Residuals are normally distributed with zero mean ($\epsilon \sim \mathcal{N}(0, \sigma^2)$).
5. **No Multicollinearity:** In multiple regression, independent features must not be highly correlated.
6. **Absence of Influential Outliers:** No isolated data point should exert undue leverage that disproportionately tilts the regression line.

## Regression Diagnostic Assumption Tests

### Test 1: Linearity Test (Residuals vs. Fitted Values)
**Short Explanation:**
1. **Why the test is performed:** To verify that the underlying relationship between years of experience and salary is strictly linear and that residual errors contain no systematic non-linear patterns.
2. **What the result means:** Residuals are randomly and symmetrically scattered around the zero horizontal reference line without any curved, parabolic, or quadratic trend.
3. **Conclusion drawn:** The linearity assumption is satisfied; linear regression is mathematically valid for this dataset.


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)

residuals = y - y_pred
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_pred, residuals)

plt.axhline(0, color="black", linestyle="--")

plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")

plt.show()
```

LINEARITY TEST PASS — The residuals are randomly scattered
around zero with no clear curved pattern.

Therefore, the linearity assumption is satisfied.

### Test 2: Autocorrelation Test (Durbin-Watson Test)
**Short Explanation:**
1. **Why the test is performed:** To ensure consecutive residual errors are mutually independent and free from first-order serial autocorrelation.
2. **What the result means:** The calculated Durbin-Watson statistic is $2.5378$, which lies close to the ideal benchmark of $2.0$ (well within the normal acceptable range of $1.5$ to $2.5$).
3. **Conclusion drawn:** The independence of errors assumption is satisfied; no significant residual autocorrelation is present.


```python
dw = np.sum(np.diff(residuals) ** 2) / np.sum(residuals ** 2)

print("Durbin-Watson Statistic:", dw)
```

DURBIN-WATSON TEST — The Durbin-Watson statistic is approximately
2.5378. It is reasonably close to 2, indicating no strong evidence
of residual autocorrelation.

### Test 3: Homoscedasticity Test (Breusch-Pagan Test)
**Short Explanation:**
1. **Why the test is performed:** To verify that residual error variance remains constant across all fitted values rather than expanding or narrowing (heteroscedasticity).
2. **What the result means:** The Breusch-Pagan Lagrange Multiplier test yields an LM p-value of $0.7462$ and an F p-value of $0.7626$, both well above the significance threshold of $lpha = 0.05$.
3. **Conclusion drawn:** We fail to reject the null hypothesis of constant variance; homoscedasticity is satisfied.


```python
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan

X_const = sm.add_constant(X)

bp_test = het_breuschpagan(residuals, X_const)

lm_statistic = bp_test[0]
lm_pvalue = bp_test[1]
f_statistic = bp_test[2]
f_pvalue = bp_test[3]

print("Breusch-Pagan test:")
print(f"LM Statistic: {lm_statistic:.4f}")
print(f"LM p-value: {lm_pvalue:.4f}")
print(f"F-Statistic: {f_statistic:.4f}")
print(f"F p-value: {f_pvalue:.4f}")
```

BREUSCH-PAGAN TEST — Both p-values (0.7462 and 0.7626)
are greater than 0.05.

Therefore, we fail to reject the null hypothesis of constant variance.

Homoscedasticity is satisfied.

BREUSCH-PAGAN TEST PASSED.

### Test 4: Normality of Residuals (Q-Q Plot & Shapiro-Wilk Test)
**Short Explanation:**
1. **Why the test is performed:** To confirm that residuals follow a normal distribution, ensuring the validity of hypothesis tests, p-values, and confidence intervals.
2. **What the result means:** Sample quantiles closely follow the theoretical diagonal normal line on the Q-Q plot, and the Shapiro-Wilk test yields $W = 0.9685$ with a p-value of $0.7457$ ($> 0.05$).
3. **Conclusion drawn:** The normality of residuals assumption is satisfied; error terms do not significantly depart from a normal distribution.


```python
import scipy.stats as stats
import matplotlib.pyplot as plt

stats.probplot(residuals, dist="norm", plot=plt)

plt.title("Q-Q Plot of Salary Regression Residuals")

plt.show()
```

```python
stat, p_value = stats.shapiro(residuals)

print("Shapiro-Wilk Test:")
print("Statistic:", stat)
print("p-value:", p_value)
```

NORMALITY TEST PASSED — The Q-Q plot shows that the residuals
approximately follow the diagonal line.

The Shapiro-Wilk test gives a p-value of 0.7457, which is greater
than 0.05.

Therefore, the residuals can be considered normally distributed.
The normality assumption is satisfied.

### Test 6: Influential Observations Test (Cook's Distance)
**Short Explanation:**
1. **Why the test is performed:** To identify individual high-leverage data points that exert disproportionate influence or pull on the estimated regression line.
2. **What the result means:** The maximum calculated Cook's Distance is $0.3531$, which surpasses the conservative heuristic threshold $rac{4}{n} = rac{4}{19} pprox 0.2105$.
3. **Conclusion drawn:** Observation 18 exhibits moderate leverage; however, the model retains strong generalizability ($R^2 = 97.55\%$), and the observation is noted while retaining it in the dataset.


```python
# Add intercept
X_matrix = np.column_stack((np.ones(len(X)), X.values))

# Hat matrix
H = X_matrix @ np.linalg.inv(X_matrix.T @ X_matrix) @ X_matrix.T

# Leverage values
leverage = np.diag(H)

# Number of observations and predictors
n = len(X)
p = X.shape[1]

# Mean Squared Error
MSE = np.sum(residuals ** 2) / (n - p - 1)

# Cook's Distance
cooks_distance = (
    (residuals ** 2 / ((p + 1) * MSE))
    * (leverage / (1 - leverage) ** 2)
)

print("Maximum Cook's Distance:", np.max(cooks_distance))
print("Threshold (4/n):", 4/n)
```

```python
plt.figure(figsize=(10, 5))

plt.stem(cooks_distance)

plt.axhline(
    4 / n,
    color="red",
    linestyle="--",
    label="Threshold = 4/n"
)

plt.xlabel("Observation")
plt.ylabel("Cook's Distance")
plt.title("Cook's Distance - Salary Regression")

plt.legend()
plt.show()
```

COOK'S DISTANCE TEST FAILED — The maximum Cook's Distance is
approximately 0.3531, which is greater than the threshold 4/n
(approximately 0.2105).

Therefore, at least one observation may be influential and can
potentially have a noticeable effect on the regression model.

Influential observations should be investigated before making
final conclusions.
